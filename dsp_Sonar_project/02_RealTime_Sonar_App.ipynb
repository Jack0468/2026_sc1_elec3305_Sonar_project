```python
# PRESET 1 — Baseline (6-12 kHz, good SNR, general purpose)
PRESET_1 = dict(f0=6000,  f1=12000, Npulse=500,  Nseg=4096, Nrep=24, Nplot=200, maxdist=200, temperature=21)

# PRESET 2 — High Resolution (wide BW=10kHz, sharper peaks, close range)
PRESET_2 = dict(f0=8000,  f1=18000, Npulse=256,  Nseg=4096, Nrep=24, Nplot=200, maxdist=150, temperature=21)

# PRESET 3 — Long Range (low freq attenuates less, long pulse for SNR gain)
PRESET_3 = dict(f0=4000,  f1=8000,  Npulse=1024, Nseg=8192, Nrep=16, Nplot=200, maxdist=400, temperature=21)

# PRESET 4 — Near-Ultrasonic (15-20 kHz, near-silent to human ear)
PRESET_4 = dict(f0=15000, f1=20000, Npulse=512,  Nseg=4096, Nrep=24, Nplot=200, maxdist=200, temperature=21)
```

```python
fs = 48000
params = PRESET_1.copy()   # ← swap to PRESET_2/3/4 as needed

input_device_index = None  # ← set to device index from the listing cell if needed

# IMPORTANT: rtsonar returns (stop_flag, win) — must unpack both
stop_flag, win = rtsonar(
    f0=params['f0'], f1=params['f1'], fs=fs,
    Npulse=params['Npulse'], Nseg=params['Nseg'],
    Nrep=params['Nrep'],     Nplot=params['Nplot'],
    maxdist=params['maxdist'], temperature=params['temperature'],
    input_device_index=input_device_index,
)
```
